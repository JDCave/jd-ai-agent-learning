#### 组件速查表

| 技术栈类别 | 解决的问题 | 代表技术与组件（2026） |
|---|---|---|
| **Vertical Agents**<br>垂直领域智能体 | 面向具体行业/场景的成品智能体 | **编程**：Cursor、Claude Code、GitHub Copilot、OpenAI Codex、Devin<br>**应用构建**：Lovable、v0、Bolt.new、Replit Agent<br>**客服/企业**：Sierra、Salesforce Agentforce、Zendesk AI<br>**垂直行业**：Harvey（法律）、Glean（企业知识） |
| **Agent Hosting & Serving**<br>托管与服务 | 智能体的部署、运行与弹性伸缩 | **云厂商托管**：AWS Bedrock AgentCore、Azure AI Foundry、Vertex AI Agent Engine<br>**厂商平台**：OpenAI AgentKit、LangGraph Platform<br>**Serverless/边缘**：Cloudflare Agents、Modal、Railway |
| **Observability**<br>可观测性与评估 | 追踪、调试与质量评估 | **追踪/监控**：LangSmith、Langfuse、Braintrust、Arize Phoenix、W&B Weave、Helicone、AgentOps<br>**标准**：OpenTelemetry GenAI<br>**评估测试**：Promptfoo、DeepEval |
| **Agent Frameworks**<br>智能体框架 | 编排模型调用、工具与控制流 | **图编排**：LangChain / LangGraph<br>**官方 SDK**：OpenAI Agents SDK、Anthropic Agent SDK、Google ADK、Microsoft Agent Framework（SK 与 AutoGen 合并）<br>**多智能体**：CrewAI、Agno、LlamaIndex<br>**轻量/类型安全**：PydanticAI、Mastra、Smolagents |
| **Memory**<br>记忆 | 跨步骤、跨会话的状态与知识 | **专用记忆层**：Mem0、Zep / Graphiti、Letta、LangMem、Supermemory、Cognee<br>**方法**：上下文工程、记忆块、Graph RAG |
| **Tool Libraries**<br>工具库与协议 | 连接外部世界：API、浏览器、其他智能体 | **协议**：MCP（工具）、A2A（智能体互联）、AG-UI（前端交互）<br>**集成平台**：Composio、Pipedream、Nango、Arcade<br>**浏览器自动化**：Browser Use、Playwright MCP、Browserbase、Firecrawl |
| **Sandboxes**<br>沙箱与执行环境 | 安全隔离地执行代码与操作 | E2B、Daytona、Modal、Cloudflare Sandboxes、Docker / Firecracker、Pyodide（WASM） |
| **Model Serving**<br>模型服务 | 提供底层模型的推理能力 | **闭源 API**：OpenAI GPT-5、Claude 4.5、Gemini 3、Grok<br>**开源模型**：Llama 4、DeepSeek、Qwen、Gemma<br>**推理/路由**：OpenRouter、LiteLLM、Together AI、Fireworks、Groq、Cerebras<br>**自托管引擎**：vLLM、SGLang、Ollama |
| **Storage**<br>存储 | 持久化向量、图与文件 | **向量库**：Pinecone、Weaviate、Qdrant、Milvus、Chroma、LanceDB<br>**数据库内置**：pgvector（默认起步）、Redis、MongoDB Atlas<br>**图数据库**：Neo4j<br>**对象存储**：S3 / R2 |

## Practice

1. 物理符号系统假说<sup>[1]</sup>是符号主义时代的理论基石。请分析：
   
   - 该假说的"充分性论断"和"必要性论断"分别是什么含义？
   - 结合本章内容，说明符号主义智能体在实践中遇到的哪些问题对该假说的"充分性"提出了挑战？
   - 大语言模型驱动的智能体是否符合物理符号系统假说？

2. 专家系统MYCIN<sup>[2]</sup>在医疗诊断领域取得了显著成功，但最终并未大规模应用于临床实践。请思考：
   
   > <strong>提示</strong>：可以从技术、伦理、法律、用户接受度等多个角度分析
   
   - 除了本章提到的"知识获取瓶颈"和"脆弱性"，还有哪些因素可能阻碍了专家系统在医疗等高风险领域的应用？
   - 如果让现在的你设计一个医疗诊断智能体，你会如何设计系统来克服MYCIN的局限？
   - 在哪些垂直领域中，基于规则的专家系统至今仍然是比深度学习更好的选择？请举例说明。

3. 在2.2节中，我们实现了一个简化版的ELIZA聊天机器人。请在此基础上进行扩展实践：
   
   > <strong>提示</strong>：这是一道动手实践题，建议实际编写代码
   
   - 为ELIZA添加3-5条新的规则，使其能够处理更多样化的对话场景（如谈论工作、学习、爱好等）
   - 实现一个简单的"上下文记忆"功能：让ELIZA能够记住用户在对话中提到的关键信息（如姓名、年龄、职业），并在后续对话中引用
   - 对比你扩展后的ELIZA与[ChatGPT](https://chatgpt.com/)，列举至少3个维度上存在的本质差异
   - 为什么基于规则的方法在处理开放域对话时会遇到"组合爆炸"问题并且难以扩展维护？能否使用数学的方法来说明？

4. 马文·明斯基在"心智社会"理论<sup>[7]</sup>中提出了一个革命性的观点：智能源于大量简单智能体的协作，而非单一的完美系统。
   
   - 在图2.6"搭建积木塔"的例子中，如果 `GRASP` 智能体突然失效了，整个系统会发生什么？这种去中心化架构的优势和劣势是什么？
   - 将"心智社会"理论与现在的一些多智能体系统（如[CAMEL-Workforce](https://docs.camel-ai.org/key_modules/workforce)、[MetaGPT](https://github.com/FoundationAgents/MetaGPT)、[CrewAI](https://github.com/crewAIInc/crewAI)）进行对比，它们之间存在哪些关联和不同之处？
   - 马文·明斯基认为智能体可以是"无心"的简单过程，然而现在的大语言模型和智能体往往都拥有强大的推理能力。这是否意味着"心智社会"理论在大语言模型时代不再适用了？
   
5. 强化学习与监督学习是两种不同的学习范式。请分析：
   
   - 用AlphaGo的例子说明强化学习的"试错学习"机制是如何工作的
   - 为什么强化学习特别适合序贯决策问题？它与监督学习在数据需求上有什么本质区别？

   - 在大语言模型的训练过程中，强化学习起到了什么关键性的作用？

6. 预训练-微调范式是现代人工智能领域的重要突破。请深入思考：
